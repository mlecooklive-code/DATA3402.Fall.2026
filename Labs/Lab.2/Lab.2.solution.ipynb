*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def check_winner(board):
# check rows
    for row in board:
        if row[0] != 0:
            winner = True
            for i in row:
                if i != row[0]:
                    winner = False
            if winner:
                return row[0]

# transpose board
    board_t = [[board[j][i] for j in range(len(board))] for i in range(len(board[0]))]

# check columns
    for row in board_t:
        if row[0] != 0:
            winner = True
            for i in row:
                if i != row[0]:
                    winner = False
            if winner:
                return row[0]

# check first diagonal
    diagonal_1 = []
    for i in range(len(board)):
        diagonal_1.append(board[i][i])
    if diagonal_1[0] != 0:
        winner = True
        for j in diagonal_1:
            if j != diagonal_1[0]:
                winner = False
        if winner:
            return diagonal_1[0]

#check second diagonal
    diagonal_2 = []
    for i in range(len(board)):
        diagonal_2.append(board[i][len(board)-1-i])
    if diagonal_2[0] != 0:
        winner = True
        for j in diagonal_2:
            if j != diagonal_2[0]:
                winner = False
        if winner:
            return diagonal_2[0]

# check draw/incomplete
    for row in board:
        for i in row:
            if i == 0:
                return -1
    return 0
````

````python
# Write your solution here
def make_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def computer_move(board, computer_player, human_player):

    # Try to win
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = computer_player

                if check_winner(board) == computer_player:
                    return True

                board[i][j] = 0

    # Try to block the human
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = human_player

                if check_winner(board) == human_player:
                    board[i][j] = computer_player
                    return True

                board[i][j] = 0

    # Otherwise take the first empty square
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = computer_player
                return True

    return False
````

````python
def play_game_computer(n):
    player_1 = 1
    player_2 = 2

    board = game_board(n)
    game_state = -1

    draw_board(board)

    while game_state == -1:

        game_input(board, player_1)
        game_state = check_winner(board)

        if game_state == -1:
            computer_move(board, player_2, player_1)
            draw_board(board)
            game_state = check_winner(board)

    return game_state
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
import copy

def score_board(board, player):
    result = check_winner(board)

    if result == player:
        return 1
    elif result == 0:
        return 0
    elif result == -1:
        return 0
    else:
        return -1
````

````python
def generate_moves(board, player, current_player, depth):

    game_state = check_winner(board)

    if depth == 0 or game_state != -1:
        return list(), score_board(board, player)

    moves = list()
    scores = list()

    for i in range(len(board)):
        for j in range(len(board[i])):

            if board[i][j] == 0:

                new_board = copy.deepcopy(board)

                make_move(new_board, current_player, i, j)

                if current_player == 1:
                    next_player = 2
                else:
                    next_player = 1

                next_moves, score = generate_moves(
                    new_board,
                    player,
                    next_player,
                    depth - 1
                )

                this_move = [(current_player, i, j)]

                moves.append(this_move)
                scores.append(score)

    return moves, scores
````

````python
def tree_search(t, depth=1):

    if isinstance(t, (int, float)):
        return t

    if len(t) == 0:
        return 0

    values = []

    for item in t:
        values.append(tree_search(item, depth + 1))

    return (max(values) + min(values)) / depth
````

````python
def pick_move(board, player, depth=5):
    moves, scores = generate_moves(board, player, player, depth)

    result = list(map(tree_search, scores))

    move_index = result.index(max(result))

    return moves[move_index][0]
````

````python
def play_game_computer_search(n, max_depth):
    player_1 = 1
    player_2 = 2

    board = game_board(n)
    game_state = -1

    draw_board(board)

    while game_state == -1:

        game_input(board, player_1)
        game_state = check_winner(board)

        if game_state == -1:
            _, row, col = pick_move(board, player_2, max_depth)
            make_move(board, player_2, row, col)

            draw_board(board)

            game_state = check_winner(board)

    return game_state
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
def computer_vs_computer(n, max_depth):
    simple_player = 1
    smart_player = 2

    board = game_board(n)
    game_state = -1

    while game_state == -1:

        # Exercise 11 computer
        computer_move(board, simple_player, smart_player)
        game_state = check_winner(board)

        if game_state == -1:

            # Exercise 12 computer
            player, row, col = pick_move(
                board,
                smart_player,
                max_depth
            )

            make_move(board, player, row, col)

            game_state = check_winner(board)

    return game_state
````

````python
def test_computers(n, max_depth):
    smart_wins = 0
    simple_wins = 0
    draws = 0

    for i in range(10):
        result = computer_vs_computer(n, max_depth)

        if result == 2:
            smart_wins += 1
        elif result == 1:
            simple_wins += 1
        else:
            draws += 1

    print("Smart wins:", smart_wins)
    print("Simple wins:", simple_wins)
    print("Draws:", draws)

    return smart_wins / 10
````
