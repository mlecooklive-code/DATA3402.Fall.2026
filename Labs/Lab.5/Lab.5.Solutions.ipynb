3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    my_matrix = []
    for i in range(n):
        row = list()
        for j in range(m):
            row.append(c)
        my_matrix.append(row)
    return matrix(my_matrix)

def zeros(n, m):
    my_matrix = []
    for i in range(n):
        row = list()
        for j in range(m):
            row.append(0)
        my_matrix.append(row)
    return matrix(my_matrix)


def ones(n, m):
    my_matrix = []
    for i in range(n):
        row = list()
        for j in range(m):
            row.append(1)
        my_matrix.append(row)
    return matrix(my_matrix)


def eye(n):
    identity = matrix(n,n)
    for i in range(n):
        for j in range(n):
            if i == j:
                identity[i][j] = 1
    return identity
            
```

```python
class matrix:
    def __init__(self, n, m=None):
        if m != None:
            self.n = n
            self.m = m
            self.matrix_M = list()
            for i in range(n):
                row = list()
                for j in range(m):
                    row.append(0)
                self.matrix_M.append(row)
        else:
            self.matrix_M = list(n)
            self.n = len(n)
            self.m = len(n[0])
            for row in n:
                if len(row) != self.m:
                    raise ValueError

    def __getitem__(self, index):
        if isinstance(index, tuple):
            if isinstance(index[0], slice) or isinstance(index[1], slice):
                rows = self.matrix_M[index[0]]
                new_matrix = []
                for row in rows:
                    new_matrix.append(row[index[1]])
                return matrix(new_matrix)
            else:
                return self.matrix_M[index[0]][index[1]]
        else:
            return self.matrix_M[index]

    def __setitem__(self, index, value):
        self.matrix_M[index[0]][index[1]] = value

    def shape(self):
        return (self.n, self.m)

    def transpose(self):
        transpose = []
        for i in range(len(self.matrix_M[0])):
            row = []
            for j in range(len(self.matrix_M)):
                row.append(self.matrix_M[j][i])
            transpose.append(row)
        return transpose

    def row(self, n):
        return self.matrix_M[n]

    def column(self, m):
        column_matrix = self.transpose()
        return column_matrix[m]

    def to_list(self):
        return self.matrix_M

    def block(self, n_0, n_1, m_0, m_1):
        new_matrix = []
        for i in range(m_0, m_1):
            row = []
            for j in range(n_0, n_1):
                row.append(self.matrix_M[i][j])
            new_matrix.append(row)
        return new_matrix

    def scalarmul(self, c):
        for i in self.matrix_M:
            for j in self.matrix_M:
                c*self.matrix_M[i][j]

    def is_same_length(self, N):
        # N = matrix(n, m)
        if self.shape() == N.shape():
            return True
        else:
            return False

    def add(self, N):
        addition_matrix = []
        if self.is_same_length(N) == True:
            for i in range(self.n):
                row = list()
                for j in range(self.m):
                    add_value = self.matrix_M[i][j] + N.matrix_M[i][j]
                    row.append(add_value)
                addition_matrix.append(row)
            return matrix(addition_matrix)
        else:
            print("Cannot add matrices")

    def sub(self, N):
        sub_matrix = []
        if self.is_same_length(N) == True:
            for i in range(self.n):
                row = list()
                for j in range(self.m):
                    add_value = self.matrix_M[i][j] - N.matrix_M[i][j]
                    row.append(add_value)
                sub_matrix.append(row)
            return matrix(sub_matrix)
        else:
            print("Cannot subtract matrices")
    

    def mat_mult(self, N):
        if self.m == N.n:
            new_matrix = []
            for i in range(self.n):
                row = list()
                for j in range(N.m):
                    total = 0
                    for k in range(self.m):
                        total += self.matrix_M[i][k]*N.matrix_M[k][j]
                    row.append(total)
                new_matrix.append(row)
            return matrix(new_matrix)
        else:
            print("Cannot multiply matrices")           


    def element_mult(self, N):
        mult_matrix = []
        if self.is_same_length(N) == True:
            for i in range(self.n):
                row = list()
                for j in range(self.m):
                    add_value = self.matrix_M[i][j] * N.matrix_M[i][j]
                    row.append(add_value)
                mult_matrix.append(row)
            return matrix(mult_matrix)
        else:
            print("Cannot multiply matrices")

    def equals(self, N):
        if self.is_same_length(N) == True:
            for i in range(self.n):
                for j in range(self.m):
                    if self.matrix_M[i][j] != N.matrix_M[i][j]:
                        return False
            return True
        else:
            return False
```
